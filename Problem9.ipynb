```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("titanic")
```

```python
# Box Plot (Age vs Gender with Survival)
plt.figure(figsize=(7,5))
sns.boxplot(x='sex', y='age', hue='survived', data=df)
plt.title("Age Distribution by Gender and Survival")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.boxplot(x='sex', y='age', data=df)
plt.title("Age Distribution by Gender")
plt.show()
```

```python
# -----------------------------------
# Observations:
# -----------------------------------
# 1. Females have higher survival rate compared to males.
# 2. Younger passengers had better chances of survival.
# 3. Males show wider variation in age and lower survival.
# 4. Age distribution shows females are slightly younger on average compared to males.
```
