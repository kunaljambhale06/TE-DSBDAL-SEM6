```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
```

```python
df = pd.read_csv(r"D:\Jupyter Notebook\Datasets\Social_Network_Ads.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
# Select Input(X) and Output(y)
```

```python
X = df[['Age', 'EstimatedSalary']]
y = df['Purchased']
```

```python
# Split Data into Training and Testing
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
# Scale Data for Better Accuracy
```

```python
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train, y_train)
```

```python
# Predict Test Data
```

```python
y_pred = model.predict(X_test)
```

```python
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)
```

```python
TN = cm[0][0]
FP = cm[0][1]
FN = cm[1][0]
TP = cm[1][1]
```

```python
print("TN =", TN)
print("FP =", FP)
print("FN =", FN)
print("TP =", TP)
```

```python
# Calculate Performance Measures
```

```python
accuracy = (TP + TN) / (TP + TN + FP + FN)
error_rate = (FP + FN) / (TP + TN + FP + FN)
precision = TP / (TP + FP)
recall = TP / (TP + FN)
```

```python
print("Accuracy =", accuracy)
print("Error Rate =", error_rate)
print("Precision =", precision)
print("Recall =", recall)
```

```python
import matplotlib.pyplot as plt
```

```python
labels = ['Accuracy', 'Error Rate', 'Precision', 'Recall']
values = [accuracy, error_rate, precision, recall]
```

```python
plt.bar(labels, values)
plt.title("Performance of Logistic Regression")
plt.ylabel("Value")
plt.show()
```
