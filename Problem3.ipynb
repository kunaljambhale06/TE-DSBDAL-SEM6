```python
print("\nSummary Statistics of Fare Grouped by Passenger Class:\n")

group_stats = df.groupby('Pclass')['Fare'].agg(
    ['mean', 'median', 'min', 'max', 'std']
)
print(group_stats)
```

```python
# ---------------------------------------------------
# PART 2
# Statistical Details of Iris Species
# ---------------------------------------------------
```

```python
iris = pd.read_csv("iris.csv")

print("\nFirst 5 Rows of Iris Dataset:\n")
print(iris.head())
```

```python
# ----------------------------------------------
# Filter by Species
# ----------------------------------------------

setosa = iris[iris['Species'] == 'Iris-setosa']
versicolor = iris[iris['Species'] == 'Iris-versicolor']
virginica = iris[iris['Species'] == 'Iris-virginica']
```

```python
# ----------------------------------------------
# Function to display stats
# ----------------------------------------------
```

```python
def show_stats(data, species_name):
    print("\n--------------------------------")
    print("Species:", species_name)
    print("--------------------------------")
    
    print("Mean:\n", data.iloc[:,1:5].mean())
    print("\nStandard Deviation:\n", data.iloc[:,1:5].std())
    print("\nPercentile (25%):\n", data.iloc[:,1:5].quantile(0.25))
    print("\nMedian (50%):\n", data.iloc[:,1:5].quantile(0.50))
    print("\nPercentile (75%):\n", data.iloc[:,1:5].quantile(0.75))
```

```python
show_stats(setosa, "Iris-setosa")
show_stats(versicolor, "Iris-versicolor")
show_stats(virginica, "Iris-virginica")
```
